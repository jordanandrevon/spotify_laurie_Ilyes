```python
### Question 4 - Artistes populaires et suivis

# Regroupement des données par artiste
# On conserve uniquement les artistes ayant au moins 3 titres
artistes_analyse = (
    df_spotify
    .groupby("artist_name")
    .agg(
        nombre_titres=("track_name", "count"),
        popularite=("artist_popularity", "max"),
        followers=("artist_followers", "max")
    )
    .query("nombre_titres >= 3")
)

artistes_analyse.head()
```

```python
# Calcul des seuils à partir des médianes
# La médiane permet de limiter l'influence des artistes ayant énormément de followers
seuil_popularite = artistes_analyse["popularite"].median()
seuil_followers = artistes_analyse["followers"].median()

# Sélection des artistes ayant une popularité et un nombre de followers
# supérieurs ou égaux aux valeurs médianes
artistes_forts = artistes_analyse[
    (artistes_analyse["popularite"] >= seuil_popularite) &
    (artistes_analyse["followers"] >= seuil_followers)
].sort_values(
    ["popularite", "followers"],
    ascending=False
)

artistes_forts
```

```python
# Sélection des 10 artistes les plus performants
# selon leur popularité et leur nombre de followers
top_artistes_forts = artistes_forts.head(10)

top_artistes_forts
```

```python
### Graphique - Popularité et nombre de followers

plt.figure(figsize=(10, 6))

# Création du nuage de points
plt.scatter(
    top_artistes_forts["followers"],
    top_artistes_forts["popularite"],
    s=100
)

# Ajout du nom de chaque artiste à côté de son point
for artiste, ligne in top_artistes_forts.iterrows():
    plt.annotate(
        artiste,
        (ligne["followers"], ligne["popularite"]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.xlabel("Nombre de followers")
plt.ylabel("Popularité de l'artiste")
plt.title("Artistes combinant popularité et nombre de followers")

# Utilisation d'une échelle logarithmique pour mieux représenter
# les écarts importants entre les nombres de followers
plt.xscale("log")

plt.show()
```

```python
### Question 5 - Caractéristiques des titres populaires

# Comparaison de la durée moyenne entre tous les titres
# et les 10 titres les plus populaires

duree_moyenne_dataset = df_spotify["track_duration_min"].mean()
duree_moyenne_top10 = top_titres["track_duration_min"].mean()

print("Durée moyenne de tous les titres :", round(duree_moyenne_dataset, 2), "minutes")
print("Durée moyenne du Top 10 :", round(duree_moyenne_top10, 2), "minutes")
```
