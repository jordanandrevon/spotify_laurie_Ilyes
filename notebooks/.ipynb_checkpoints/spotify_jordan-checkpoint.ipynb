```python
#### Top 10 des titres les plus populaires 
```

```python
top_titres = df_spotify.sort_values(
    by="track_popularity",
    ascending=False
).head(10)

top_titres[[
    "track_name",
    "artist_name",
    "album_name",
    "track_popularity"
]]

top_titres[[
    "track_name",
    "artist_name",
    "artist_genres",
    "album_name",
    "track_duration_min",
    "track_popularity"
]]
```

```python
#### Top 10 des titres les moins populaires 
```

```python
flop_titres = df_spotify.sort_values(
    by="track_popularity",
    ascending=True
).head(10)

flop_titres[[
    "track_name",
    "artist_name",
    "album_name",
    "track_popularity"
]]

flop_titres[[
    "track_name",
    "artist_name",
    "artist_genres",
    "album_name",
    "track_duration_min",
    "track_popularity"
]]
```

```python
### Top Artistes

top_artistes = (
    df_spotify
    .groupby("artist_name")
    .agg(
        nombre_titres=("track_name", "count"),
        popularite_moyenne=("track_popularity", "mean"),
        followers=("artist_followers", "max")
    )
    .query("nombre_titres >= 3")
    .sort_values("popularite_moyenne", ascending=False)
    .head(10)
)

top_artistes
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    top_titres["track_name"],
    top_titres["track_popularity"]
)

plt.xlabel("Popularité")
plt.ylabel("Titre")
plt.title("Top 10 des titres les plus populaires")
plt.gca().invert_yaxis()

plt.show()
```

```python
####Quels sont les artistes ayant la meilleure popularité moyenne parmi ceux qui possèdent plusieurs titres dans notre dataset ?
top_artistes = (
    df_spotify
    .groupby("artist_name")
    .agg(
        nombre_titres=("track_name", "count"),
        popularite_moyenne=("track_popularity", "mean"),
        followers=("artist_followers", "max")
    )
    .query("nombre_titres >= 3")
    .sort_values("popularite_moyenne", ascending=False)
    .head(10)
)

top_artistes
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    top_artistes.index,
    top_artistes["popularite_moyenne"]
)

plt.xlabel("Popularité moyenne")
plt.ylabel("Artiste")
plt.title("Top 10 des artistes selon leur popularité moyenne")
plt.gca().invert_yaxis()

plt.show()
```

```python
top_followers = (
    df_spotify
    .groupby("artist_name")
    .agg(
        followers=("artist_followers", "max"),
        popularite_moyenne=("track_popularity", "mean"),
        nombre_titres=("track_name", "count")
    )
    .query("nombre_titres >= 3")
    .sort_values("followers", ascending=False)
    .head(10)
)

top_followers
```
